```python
import tensorflow as tf
x=tf.constant([1.0,2.0,3.0,4.0])
print("Square:",tf.math.square(x))
print("Square root:",tf.math.sqrt(x))
print("Exponential:",tf.math.exp(x))
print("Logarithm:",tf.math.log(x))
```

```python
import tensorflow as tf
x=tf.constant([10.0,20.0,30.0,40.0])
print("Sum:",tf.reduce_sum(x))
print("Mean:",tf.reduce_mean(x))
print("Maximum:",tf.reduce_max(x))
print("Minimum:",tf.reduce_min(x))
```

```python
x=tf.constant([[10,20,30],[40,50,60]])
print(x)
print(tf.reduce_sum(x))
print(tf.reduce_sum(x, axis=1))
print(tf.reduce_sum(x, axis=0))
```

```python
import tensorflow as tf
x=tf.constant([[10,20,30],[40,50,60]])
print("Original:")
print(x)
print("Sum axis=0:")
print(tf.reduce_sum(x,axis=0))
print("Sum axis=1:")
print(tf.reduce_sum(x,axis=1))
print("Mean axis=0:")
print(tf.reduce_mean(x,axis=0))
print("Mean axis=1:")
print(tf.reduce_mean(x,axis=1))
```
