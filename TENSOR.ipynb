```python
#what is a tensors
# a tensor is a multidimensional array of numbers
#scalar=0D tensor -- a single number
#[1,2,3] --vector 1D tensor
#why do we need tensors?  
```

```python
#creating tensors in PYTORCH
#ndim tells dimensions
import torch
x=torch.tensor([1,2,3])
print(x)
print(x.ndim)
b=torch.tensor([[1,2,3],[4,5,6]])
print(b.ndim)
```

```python
#shape tells how many elements exist alond each dimension
x=torch.tensor([1,2,3])
b=torch.tensor([[1,2,3],[4,5,6]])
print(b.shape)
```

```python

x=torch.tensor([1,2,3])
b=torch.tensor([[1,2,3],[4,5,6]])
print(b.size())
```

```python
#numel tells how many elements exist 
b=torch.tensor([[1,2,3],[4,5,6]])
print(b.numel())
```

```python
#tensor datatype
b=torch.tensor([[1,2,3],[4,5,6]])
print(b.dtype)
```

```python
x=torch.tensor([1.,2.,3.])
print(x.dtype)
print(x.device)
```

```python

v=torch.zeros(3)
print(v)
b=torch.zeros(2,3)
print(b)
```

```python

v=torch.ones(3)
print(v)
b=torch.ones(2,3)
print(b)
```

```python

v=torch.rand(3)
print(v)
b=torch.rand(2,3)
print(b)
```

```python

v=torch.randn(3)
print(v)
b=torch.randn(3,3)
print(b)
```

```python
# creating IDENTY MATRIX
v=torch.eye(3)
print(v)
b=torch.eye(4)
print(b)
```

```python

v=torch.arange(3)
print(v)
b=torch.arange(0,10)
print(b)
n=torch.arange(0,10,3)
print(n)
```

```python
#evenly spaced values between two numbers
b=torch.linspace(0,5,4)
print(b)
```

```python
torch.manual_seed(42)
torch.rand(3)
```

```python
#indexing tensors
x=torch.tensor([1,2,3,4,5,6])
x[0]
x[5]
```

```python
#2D indexing
x=torch.tensor([[12,3,4,5],[2,45,89,4]])
print(x)
x[1][1]
x[1][3]
x[1,2]
```

```python
#slicing
x=torch.tensor([10,20,30,40,50])
x[1:4]
r=torch.tensor([[12,3,4,5],[2,45,89,4]])
print(r)
r[1:,1:2]
r[0:2,2:4]

```

```python
#tensor operations
a=torch.tensor([1,2,34,45])
b=torch.tensor([7,8,9,10])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
#scalar operations
x=torch.tensor([1,3,4,5])
print(x+10)
print(x*10)
print(x-10)
```

```python
#Matrix multiplication
# here element wise multiplication (a*b) is different from the matrix multiplication (a@b) or(matmul(a,b))
a=torch.tensor([[1,2],[3,4]])
b=torch.tensor([[3,4],[4,8]])
print(a@b)
```

```python
#Aggregation operations
x=torch.tensor([1,3,4,5])
print(x.sum())
print(x.min())
print(x.max())
print(x.std)
```

```python
x=torch.tensor([1,3,4,5])
print(x.argmax())#index value
```

```python
x=torch.arange(1,19)
print(x)
print(x.shape)
print(x.reshape(3,6))
print(x.reshape(2,9))
print(x.reshape(2,3,3))
```

```python
x=torch.arange(1,9)
print(x)
print(x.view(3,3))
```

```python
x=torch.tensor([1,2,3])
print(x.shape)
t=x.unsqueeze(0)
print(t.shape)
f=x.unsqueeze(1)
print(f.shape)
```

```python
#transpose
x=torch.tensor([[1,2,3],[4,5,6]])
print(x)
print(x.T)
```

```python
#CONCATENATION
a=torch.tensor([1,2])
b=torch.tensor([3,4])
print(torch.cat((a,b)))
```

```python
#Broadcasting
x=torch.tensor([[1,2,3],[4,5,6]])
y=torch.tensor([10,20,30])
print(x+y)
```
