```python
#loc --[label-based selection]
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
print(df)
print("\nhere loc[0]  gives information of the index of 0")
print(df.loc[0])
```

```python
#loc --[label-based selection]
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
print(df)
print("\nhere loc[0]  gives information of selected rows") 
print(df.loc[[0,2]])
```

```python
#iloc --[integer position based selection]
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
print(df)
print("\nhere loc[0]  gives information of the index of 0")
print(df.iloc[0])
print("\n")
print(df.iloc[[0,2]])
```

```python
#seleccting a particular cell using loc , iloc
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
print(df)
print("\nhere loc[0,2]  gives information of the row 0 col 1")
print(df.loc[1,"marks"])
print("\n")
print(df.iloc[1,2])
```

```python
#FILTERING DATA
#suppose we want the marks greater than 80
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[79,89,98]})
print(df)
print("\n")
print("marks greater than 80")
print(df[df["marks"]>80])

```

```python
#for MULTIPLE CONDITIONS
#suppose age>18 and marks >80
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
print(df)
print("\n")
print(df[(df["age"]>18)&(df["marks"]>80)])
print("\n")
print(df[(df["age"]>18)|(df["marks"]>80)])
```

```python
#adding a new coloumn
#suppose we want result colom
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
df["result"]="pass" # here result colom is added
print(df)
print("\n we acn also calculate a col")
df["bonus"]=df["marks"]+5
print(df)
```

```python
#removing a coloumn
#suppose we want result colom
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
df["result"]="pass" # here result colom is added
print(df)
print("\n we acn also calculate a col")
df["bonus"]=df["marks"]+5
print(df)
#df=df.drop("bonus",axis=1)
#print("\n after removing bonus col")
#print(df)
df=df.drop("bonus",axis=1,inplace=True)
print(df)

```

```python
#sorting
#suppose we want students sorted by marks
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,18,19],"marks":[87,89,98]})
df["result"]="pass" # here result colom is added
print(df)
c=df.sort_values("marks")
d=df.sort_values("marks",ascending=False)
print("\n ascended by marks")
print(c)
print("\n decended by marks")
print(d)
```

```python
#missing values
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,pd.NaT,19],"marks":[87,pd.NA,pd.NaT]})
print(df)
print(df.isnull())
print(df.isnull().sum())
```

```python
#removing missing values
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,pd.NaT,19],"marks":[87,pd.NA,pd.NaT]})
print(df.dropna())
```

```python
#filling missing values
#instead of removing them we can replace them
import pandas as pd
df=pd.DataFrame({"name":["ravi","sita","arun"],"age":[18,pd.NaT,19],"marks":[87,pd.NA,pd.NaT]})
print(df.fillna(0))
```
