```python
#Ticks
#xticks are the numbers / [positions shown along x-axis
#yticks are the numbers / [positions shown along y-axis
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.xticks([0,2,4,6,8,10])
plt.yticks([0,20,40,60,80,100])
plt.show()
```

```python
#Ticks
#xticks are the numbers / [positions shown along x-axis
#yticks are the numbers / [positions shown along y-axis
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.subplot(2,2,1)
plt.xticks([0,2,4,6,8,10],["sri","valli","latha","laxmi","musili","fog"])
plt.yticks([0,20,40,60,80,100])
plt.subplot(2,2,2)
plt.scatter(x,y)
plt.subplot(2,2,4)
plt.bar(x,y)
plt.subplot(2,2,1)
plt.hist(x,y)
plt.show()
```

```python
import matplotlib.pyplot as plt
rk=[8,7,13,45,67,3]
tier1=["IIT Dhonbad(jharkahand)","IIT Roorkee(utharkhand)","IIT madras(chennai)","IIT KHARGAPUR(west bangel)","IIT guwahat(assam)","IIT hyderabad(telangana)"]
tier2=["IIIT DM(kurnool)","IIIT kalyani(west bangal)","IIIT lucknow","IIIT sricity(A.P)","IIIT ONGOLE(A.P)","IIIT hyderabad(telangana)"]
plt.plot(tier1,rk,marker="o",linestyle="--",color="green",label="IIT COLLEGES")
plt.xlabel("IIT Colleges")
plt.ylabel("Rank")
plt.title("IIT Colleges vs Rank")
plt.xticks(rotation=45, ha="right")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
rk=[8,7,13,45,67,3]
tier1=["IIT Dhonbad(jharkahand)","IIT Roorkee(utharkhand)","IIT madras(chennai)","IIT KHARGAPUR(west bangel)","IIT guwahat(assam)","IIT hyderabad(telangana)"]
tier2=["IIIT DM(kurnool)","IIIT kalyani(west bangal)","IIIT lucknow","IIIT sricity(A.P)","IIIT ONGOLE(A.P)","IIIT hyderabad(telangana)"]
plt.plot(tier1,rk,marker="o",linestyle="--",color="green",label="IIT COLLEGES")
plt.plot(tier2, rk, marker="s", linestyle="--", label="IIIT Colleges")
plt.xlabel("Colleges")
plt.ylabel("Rank")
plt.title("IITs vs IIITs Rank Comparison")
plt.xticks(rotation=90, ha="right")
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend()

plt.show()
```
